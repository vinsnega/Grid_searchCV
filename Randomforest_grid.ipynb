```python
import pandas as pd
```

```python
dataset=pd.read_csv("insurance_pre.csv")
dataset
```

```python
dataset=pd.get_dummies(dataset,drop_first=True)
```

```python
Independent=dataset[['age', 'bmi', 'children','sex_male', 'smoker_yes']]
dependent=dataset[['charges']]
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,Y_train,Y_test=train_test_split(Independent,dependent, test_size=0.30,random_state=0)
```

```python
from sklearn.preprocessing import StandardScaler
sc=StandardScaler()
X_train=sc.fit_transform(X_train)
X_test=sc.transform(X_test)
```

```python
from sklearn.model_selection import GridSearchCV
from sklearn.ensemble import RandomForestRegressor
para_grid={'criterion':['squared_error','absolute_error','poisson'],
        
           'max_features':['sqrt','log2','None'],
           'n_estimators':[10,100]}
grid=GridSearchCV(RandomForestRegressor(),para_grid, refit=True, verbose=3,n_jobs=-1)
grid.fit(X_train,Y_train)
```

```python
re=grid.cv_results_
grid_pred=grid.predict(X_test)
from sklearn.metrics import r2_score
r_score=r2_score(Y_test,grid_pred)
print("The R2 score value for the best parameter{}:".format(grid.best_params_),r_score)
```

```python
table=pd.DataFrame.from_dict(re)
table
```

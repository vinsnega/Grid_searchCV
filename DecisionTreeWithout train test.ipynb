```python
import pandas as pd
```

```python
dataset=pd.read_csv("insurance_pre.csv")
dataset=pd.get_dummies(dataset,drop_first=True)
dataset
```

```python
Independent=dataset[['age', 'bmi', 'children','sex_male', 'smoker_yes']]
dependent=dataset[['charges']]
```

```python
from sklearn.model_selection import GridSearchCV
from sklearn.tree import DecisionTreeRegressor
para_grid={'criterion':['squared_error','absolute_error','poisson'],
           'max_features': ['auto','sqrt','log2'],
           'splitter':['best','random']}
grid=GridSearchCV(DecisionTreeRegressor(),para_grid,refit= True, verbose=3,n_jobs=-1)
grid.fit(Independent,dependent)
```

```python
re=grid.cv_results_
```

```python
table=pd.DataFrame.from_dict(re)
table
```

```python
age_inp=float(input("Age:"))
bmi_inp=float(input("BMI:"))
children_inp=int(input("Children:"))
sex_male_inp=int(input("Sex 0 or 1:"))
smoker_yes_inp=int(input("Smoker Yes 0 or 1:"))
```

```python
future_pred=grid.predict([[age_inp,bmi_inp,children_inp,sex_male_inp,smoker_yes_inp]])
print("Future_prediction={}".format(future_pred))
```
